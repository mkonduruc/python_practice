```python
# Question 43: Create a dictionary with nested lists: {'matrix': [[1, 2], [3, 4]], 'vector': [5, 6, 7]}

input_dict = {'matrix': [[1, 2], [3, 4]], 'vector': [5, 6, 7]}

input_dict
```

```python
# Question 44: Find sum of each nested list
input_dict = {'matrix': [[1, 2], [3, 4]], 'vector': [5, 6, 7]}

output_dict = {}

for key, value in input_dict.items():
     if all(isinstance(in_value,int) for in_value in value):
          output_dict[key] = sum(value)
     else:
          output_dict[key] = [sum(in_list) for in_list in value]     

output_dict
```

```python
# Question 49: Create a dictionary with nested sets: {'groups': {{1, 2, 3}, {4, 5, 6}}, 'categories': {{'a', 'b'}, {'c', 'd'}}}

input_dict = {'groups': {frozenset({1, 2, 3}), frozenset({4, 5, 6})}, 'categories': {frozenset({'a', 'b'}), frozenset({'c', 'd'})}}

input_dict
```

```python
# Question 50: Find union of all nested sets

result = {}

for key, values in input_dict.items():
    result[key] = set().union(*values)

print(result)
```
