```python
#1. Count how many vowels are in a given string.

input_value = 'This is an example'

vowels_count = 0

for chars in input_value:
    if chars.lower() in ('a','e','i','o','u'):
        vowels_count += 1

print(f"Number of vowels in '{input_value}': {vowels_count}")        
```

```python
#2. Reverse a number using only a for loop (no slicing).
#Example: Input → 12345, Output → 54321

input_number = 12345

value_to_check = int(input_number)
digit_count = len(str(input_number))
reversed_digits = 0

for iter in range(digit_count):
    reversed_digits = (reversed_digits*10) + (value_to_check % 10)
    value_to_check //= 10

print(f"Reversed digits of {input_number} is {int(reversed_digits)}")
```

```python
#3. Find the sum of digits of a number.
#Example: 5483 → 20

input_number = 5483

value_to_check = int(input_number)
sum_digits = 0

while value_to_check > 0:
    sum_digits += value_to_check % 10
    value_to_check //= 10

print(f"Reversed digits of {input_number} is {int(sum_digits)}")
```

```python
#4. Print all even numbers between 1 and 100 using a for loop.

print ("Even numbers between 1 and 100")

for number in range(1,101):
    if number%2 == 0:
        print(number)
```

```python
#5. Find the factorial of a number.

factorial_value = 5
result = 1

for number in range(factorial_value, 0 , -1):
    result *= number 

print(f"Factorial of {factorial_value} is {result}")
```

```python
#6. Check whether a number is prime using a for loop.

input_number = 23

is_prime = "is a Prime number"

for number in range(2,input_number,1):
    if input_number%number == 0:
        is_prime = "is not a Prime number"

print(f"Number {input_number} {is_prime}")
```

```python
#7. Print the multiplication table of a number up to 10.

table = 10

for second_number in range(1,11):
    print(f"{table} X {second_number} = {table*second_number}")
```

```python
#8. Count how many digits are in a number.
input_number = 12345

value_to_check = int(input_number)
digit_count = 0

while value_to_check > 0:
    value_to_check //= 10
    digit_count += 1

print(f"{input_number} has {digit_count} digits")
```

```python
#9. Find the largest number in a list.
#Example: [4, 7, 1, 9, 3]

input_list = [4, 7, 1, 9, 3]
max_number = 0

for number in input_list:
    if max_number < number:
        max_number = number

print(f"Largest number in the list {input_list} = {max_number}")
```

```python
#11. From 1 to 50, find:
#a) sum of even numbers
#b) sum of odd numbers

sum_evens = 0
sum_odds = 0

for number in range(1,51):
    if number%2 == 0:
        sum_evens += number
    else:
        sum_odds += number

print(f"sum of even numbers and sum of odd numbers from 1 to 50 is: {sum_evens} and {sum_odds}")            
```

```python
#12. Remove all spaces from a string.
#Example: "data science rocks" → "datasciencerocks"

input_value = "data science rocks"

for input in input_value:
    if input.isalpha():
        print(input, end='')
```

```python
#13. Check whether a string is a palindrome.
#Example: "madam"

input_value = 'madam'
reverse_value = ''

for str in range(len(input_value),0,-1):
    reverse_value += input_value[str-1]

print(f"{input_value} is a palindrom: {input_value == reverse_value}")    
```

```python
#14. Find duplicate values in a list.
#Example: [1,2,3,2,4,5,1]

input_list = [1,2,3,2,4,5,1]

duplicate_list = []

for number in input_list:
    if input_list.count(number) > 1 and duplicate_list.count(number) == 0:
        duplicate_list.append(number)

print(f"Duplicate values in {input_list}: {duplicate_list}")
```
