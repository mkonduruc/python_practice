## Question 1: Bank Account System

### Create a class `BankAccount` with the following:

#### Attributes:
- `name` (account holder name)
- `balance` (initial balance)

#### Methods:
- `deposit(amount)`  
  Adds money to balance

- `withdraw(amount)`  
  Subtracts money if balance is sufficient, otherwise print `"Insufficient balance"`

- `display_balance()`  
  Prints current balance

---

### Requirements:
- Maintain a transaction history using a list
- Use a loop to allow the user to perform multiple operations (deposit, withdraw, exit)
- Validate input (amount should be positive)

```python
class BankAccount:
    def __init__(self, name, balance=0.0):
        if not isinstance(name, str) or not name.strip():
            raise ValueError("Name must be a non-empty string.")
        if not isinstance(balance, (int, float)) or balance < 0:
            raise ValueError("Initial balance must be a non-negative number.")
        
        self.name = name.strip()
        self.balance = float(balance)

    def deposit(self, amount):

        if not isinstance(amount, (int, float)) or amount <= 0:
            print("Deposit amount must be a positive number.")
            return
        self.balance += amount
        print(f"Deposited ₹{amount:.2f}. New balance: ₹{self.balance:.2f}")

    def withdraw(self, amount):

        if not isinstance(amount, (int, float)) or amount <= 0:
            print("Withdrawal amount must be a positive number.")
            return
        if amount > self.balance:
            print("Insufficient balance")
        else:
            self.balance -= amount
            print(f"Withdrew ₹{amount:.2f}. New balance: ₹{self.balance:.2f}")

    def __str__(self):

        return f"Account Holder: {self.name}, Balance: ₹{self.balance:.2f}"


# Example usage:
if __name__ == "__main__":
    account = BankAccount("Alice", 5000)

    print(account)  # Show initial details

    account.deposit(1500)   # Deposit money
    account.withdraw(2000)  # Withdraw money
    account.withdraw(6000)  # Attempt to withdraw more than balance

    print(account)  # Final account details
```

## Question 2: Student Management System

### Create a class `Student` with:

#### Attributes:
- `name`
- `marks` (dictionary with subject as key and marks as value)

#### Methods:
- `calculate_total()`  
  Returns total marks

- `calculate_average()`  
  Returns average marks

- `get_grade()`  
  - average >= 90 → Grade A  
  - average >= 75 → Grade B  
  - average >= 50 → Grade C  
  - else → Fail  

---

### Requirements:
- Store multiple student objects in a list
- Use a loop to display each student’s name, total, average, and grade
- Handle empty marks dictionary case


```python
class Student:
    def __init__(self, name, marks):

        if not isinstance(name, str) or not name.strip():
            raise ValueError("Name must be a non-empty string.")
        if not isinstance(marks, dict) or not marks:
            raise ValueError("Marks must be a non-empty dictionary.")
        for subject, score in marks.items():
            if not isinstance(subject, str) or not subject.strip():
                raise ValueError("Subject names must be non-empty strings.")
            if not isinstance(score, (int, float)) or score < 0 or score > 100:
                raise ValueError("Marks must be numbers between 0 and 100.")

        self.name = name.strip()
        self.marks = marks

    def calculate_total(self):
        return sum(self.marks.values())

    def calculate_average(self):
        return self.calculate_total() / len(self.marks)

    def get_grade(self):
        avg = self.calculate_average()
        if avg >= 90:
            return "Grade A"
        elif avg >= 75:
            return "Grade B"
        elif avg >= 50:
            return "Grade C"
        else:
            return "Fail"


# Example usage
if __name__ == "__main__":
    try:
        # Example student data
        student1 = Student("Alice", {"Math": 95, "Science": 88, "English": 92})
        print(f"Student: {student1.name}")
        print(f"Total Marks: {student1.calculate_total()}")
        print(f"Average Marks: {student1.calculate_average():.2f}")
        print(f"Result: {student1.get_grade()}")

        student2 = Student("Bob", {"Math": 60, "Science": 55, "English": 50})
        print(f"\nStudent: {student2.name}")
        print(f"Total Marks: {student2.calculate_total()}")
        print(f"Average Marks: {student2.calculate_average():.2f}")
        print(f"Result: {student2.get_grade()}")

    except ValueError as e:
        print(f"Error: {e}")
```

## Question 3: Library Management System

### Create two classes:

#### Class `Book`

##### Attributes:
- `title`
- `author`
- `is_available` (default True)

---

#### Class `Library`

##### Attributes:
- List of books

##### Methods:
- `add_book(book)`  
  Adds a book to library  

- `borrow_book(title)`  
  Marks book as not available if found, else print appropriate message  

- `return_book(title)`  
  Marks book as available  

- `display_books()`  
  Prints all books with availability status  

---

### Requirements:
- Use loops to search for books
- Use conditionals to check availability
- Handle case when book is not found


```python
class Book:
    def __init__(self, title, author, is_available=True):
        if not title or not author:
            raise ValueError("Title and author cannot be empty.")
        self.title = title
        self.author = author
        self.is_available = is_available

    def __str__(self):
        status = "Available" if self.is_available else "Not Available"
        return f"'{self.title}' by {self.author} - {status}"


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        if not isinstance(book, Book):
            print("Error: Only Book instances can be added.")
            return
        self.books.append(book)
        print(f"Book '{book.title}' added to the library.")

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.is_available:
                    book.is_available = False
                    print(f"You have borrowed '{book.title}'.")
                else:
                    print(f"Sorry, '{book.title}' is currently not available.")
                return
        print(f"Book titled '{title}' not found in the library.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.is_available:
                    book.is_available = True
                    print(f"Book '{book.title}' has been returned.")
                else:
                    print(f"Book '{book.title}' was not borrowed.")
                return
        print(f"Book titled '{title}' not found in the library.")

    def display_books(self):
        if not self.books:
            print("No books in the library.")
            return
        print("\nLibrary Books:")
        for book in self.books:
            print(f" - {book}")


# Example usage
if __name__ == "__main__":
    library = Library()

    # Adding books
    library.add_book(Book("The Great Gatsby", "F. Scott Fitzgerald"))
    library.add_book(Book("1984", "George Orwell"))
    library.add_book(Book("To Kill a Mockingbird", "Harper Lee"))

    # Display all books
    library.display_books()

    # Borrow a book
    library.borrow_book("1984")
    library.borrow_book("1984")  # Trying to borrow again

    # Return a book
    library.return_book("1984")
    library.return_book("1984")  # Trying to return again

    # Display updated list
    library.display_books()
```
