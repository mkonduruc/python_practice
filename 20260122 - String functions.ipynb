```python
# Write a program to take a string as input and print its length.

input_string = 'Test Program'

print(f"Length of the input value:  {len(input_string)}")
```

```python
# Given a string, convert it to uppercase and lowercase.

input_string = 'Test Program'

print(f"Lower case of the input value:  {input_string.lower()}")
```

```python
# Write a program to count the number of vowels in a string.

input_string = 'Test Program'


print(f"Number of vowels in the input value:  {input_string.lower().count('a')+
                                               input_string.lower().count('e')+
                                               input_string.lower().count('i')+
                                               input_string.lower().count('o')+
                                               input_string.lower().count('u')}")
```

```python
# Check whether a given string is a palindrome (example: madam)

input_string = 'Madam'

round(len(input_string)/2)


print(f"Input value is Palindrome:  {input_string[:round(len(input_string)/2)].lower() == (input_string[-round(len(input_string)/2):])[::-1]}")
```

```python
# Given a sentence, count the number of words in it.

input_string = 'This is a Test Program'


print(f"Total number of words in Input value:  {len(input_string.split())}")
```

```python
# Replace all spaces in a string with underscores (_)

input_string = 'This is a Test Program'

print(f"Input value spaces replaced with (_):  {(input_string.replace(' ','_'))}")
```

```python
#Write a program to reverse a string without using built-in reverse functions.

input_string = 'Test Program'

print(f"Reverse of the Input value:  {(input_string[::-1])}")
```

```python
# Given a string, remove leading and trailing spaces.

input_string =  "   Learning Python is simple and Python is powerful   "

print(f"Input value leading & trailing spaces removed: '{input_string.strip()}'")
```

```python
# Check whether a string starts with a given character.

input_string =  "   Learning Python is simple and Python is powerful   "

starting_string = "P"

print(f"Input value is starting with '{starting_string}': {input_string.startswith(starting_string)}")
```

```python
# Find the index of the first occurrence of a character in a string.

input_string =  "   Learning Python is simple and Python is powerful   "

index_string = "P"

print(f"Character '{index_string}' first position in Input value: {input_string.index(index_string)}")
```

```python
# Count how many times a specific character appears in a string

input_string =  "   Learning Python is simple and Python is powerful   "

index_string = "P"

print(f"Character '{index_string}' count in Input value: {input_string.count(index_string)}")
```

```python
# Split a sentence into a list of words.

input_string =  "   Learning Python is simple and Python is powerful   "


print(f"Spliting Input value into words: {input_string.split()}")
```

```python
# Join a list of words into a single string with spaces.

input_string = ['Learning', 'Python', 'is', 'simple', 'and', 'Python', 'is', 'powerful']

print(f"Joining Input values into single string with space: {" ".join(input_string)}")
```

```python
# Replace a word in a sentence with another word

input_string =  "   Learning Python is simple and Python is powerful   "

old_word = "Python"

new_word = "SQL"

print(f"Replacing '{old_word}' to '{new_word}' in Input value: {input_string.replace(old_word,new_word)}")
```
