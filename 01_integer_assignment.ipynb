```python
# Question 1: Calculate the product of first 10 natural numbers
# print("Question 1: Calculate the product of first 10 natural numbers")

product_of_first_10_numbers = 1
numbers_list = []
for number in range(1,11):
    product_of_first_10_numbers *= number
    numbers_list.append(number)

print(f"Product of first 10 natural numbers {numbers_list} is {product_of_first_10_numbers}")
```

```python
# Question 2: Find the remainder when 156 is divided by 7
# print("\nQuestion 2: Find the remainder when 156 is divided by 7")
dividend = 156
divisor = 7
dividision = dividend/divisor
reminder = dividend%divisor
quotient = dividend//divisor

print(f"Divison of number {dividend} with {divisor} is {dividision} with a quotient {quotient} and reminder {reminder}")
```

```python
# Question 3: Calculate the square of 25
# print("\nQuestion 3: Calculate the square of 25")
input_number = 25
square_of_number = input_number**2
print(f"Square of Number: {input_number} is {square_of_number}")
```

```python
# Question 4: Find the cube root of 125
# print("\nQuestion 4: Find the cube root of 125")

input_number = 125
cube_root_of_number = input_number**(1/3)
print(f"Cube root of Number: {input_number} is {cube_root_of_number}")
```

```python
# Question 5: Calculate the sum of digits in number 12345
# print("\nQuestion 5: Calculate the sum of digits in number 12345")

number = 12345
number_list = []
sum_of_digits = 0

while number>=1:
    digit=number%10
    number = int(number/10)
    number_list.append(digit)
    sum_of_digits += digit 

number_list.reverse()

print(f"Sum of the digits in number: 12345 is {sum_of_digits} and digit list is {number_list}")
```

```python
# Question 6: Check if 97 is a prime number
# print("\nQuestion 6: Check if 97 is a prime number")

input_number = 97

is_prime = "is a Prime number"

for number in range(2,input_number,1):
    if input_number%number == 0:
        is_prime = "is not a Prime number"

print(f"Number {input_number} {is_prime}")
```

```python
# Question 7: Find the factorial of 8
# print("\nQuestion 7: Find the factorial of 8")

input_number = 8

factotial_value = 1

for number in range(input_number,1,-1):
    factotial_value *= number 

print(f"Factorial of number {input_number} is {factotial_value}")
```

```python
# Question 8: Calculate the average of numbers: 15, 23, 31, 42, 56
# print("\nQuestion 8: Calculate the average of numbers: 15, 23, 31, 42, 56")

number_list = [15, 23, 31, 42, 56]

sum_of_numbers = 0

for number in number_list:
    sum_of_numbers += number

averga_of_numbers = sum_of_numbers/len(number_list)

print(f"Average of numbers {number_list} is {averga_of_numbers}")
```

```python
# Question 9: Find the greatest common divisor (GCD) of 48 and 36
# print("\nQuestion 9: Find the greatest common divisor (GCD) of 48 and 36")

input_number1 = 48
input_number2 = 36

if input_number1 < input_number2:
    GCD_calc_start = input_number1
else: 
    GCD_calc_start = input_number2    

for number in range(1, GCD_calc_start, 1):
    if input_number1%number == 0 and input_number2%number == 0:
        GCD_Value = number

print(f"The greatest common divisor (GCD) of {input_number1} and {input_number2} is {GCD_Value}")
```

```python
# Question 10: Calculate the sum of first 20 odd numbers
# print("\nQuestion 10: Calculate the sum of first 20 odd numbers")

summ_of_20_odds = 0

for count in range(20):
    summ_of_20_odds += count + (count+1)
    print(f"{count} : {count+1} = {count + (count+1)}")

print(f"The sum of the first 20 odd numbers is {summ_of_20_odds}")
```
