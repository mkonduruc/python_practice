```python
# Question 1: Calculate the average of 3.14, 2.718, 1.618, 0.577
# print("Question 1: Calculate the average of 3.14, 2.718, 1.618, 0.577")

number_list = [3.14, 2.718, 1.618, 0.577]

sum_of_numbers = 0

for number in number_list:
    sum_of_numbers += number

averga_of_numbers = sum_of_numbers/len(number_list)

print(f"Average of numbers {number_list} is {averga_of_numbers}")
```

```python
# Question 2: Convert 98.6 Fahrenheit to Celsius (F = C * 9/5 + 32)
# print("\nQuestion 2: Convert 98.6 Fahrenheit to Celsius")

fahrenheit_input = float(input("Enter a value for calicultion: "))

celsius_output = (fahrenheit_input - 32) * (5/9)

print(f"{fahrenheit_input} Fahrenheit is equaivalent to {celsius_output} Celsius")
```

```python
# Question 3: Calculate the compound interest on $1000 at 5.5% for 3 years
# print("\nQuestion 3: Calculate compound interest on $1000 at 5.5% for 3 years")

principle_amount = 1000
rate_of_interest = 5.5
number_of_years = 3

Accured_amount = principle_amount*((1.0+((rate_of_interest/100)/12))**(number_of_years*12))

compound_interest = Accured_amount - principle_amount

print(f"compound interest on $1000 at 5.5% for 3 years is {compound_interest}")
```

```python
# Question 4: Find the hypotenuse of a right triangle with sides 3.5 and 4.2
# print("\nQuestion 4: Find the hypotenuse of a right triangle with sides 3.5 and 4.2")

side1 = 3.5
side2 = 4.2

hypotenuse = ((side1**2) + (side2**2))**(1/2)

print(f"Hypotenuse of a right triangle with sides 3.5 and 4.2 is {hypotenuse}")
```

```python
# Question 5: Calculate the volume of a sphere with radius 7.8
# print("\nQuestion 5: Calculate the volume of a sphere with radius 7.8")

raduis = 7.8
volume_sphere = (4/3)*3.14159*(raduis**3)

print(f"Volume of a sphere with radius {raduis} is {volume_sphere}")
```
