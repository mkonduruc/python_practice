```python
## Password strenght checker 


password= input("Enter paswword with atleast 8 charecters: should contain one small, one capital letters, two numbers and one symbol")

length = 0 
lower_case = 0
upper_case = 0
number =0
special_character =0

for charectrrs in password:
    length += 1
    if charectrrs.islower():
        lower_case += 1
    if charectrrs.isupper():
        upper_case += 1
    if charectrrs.isnumeric():
        number += 1
    if charectrrs in ('!','@','#','$','%','&','*'):
        special_character += 1

print(length) 
print(lower_case) 
print(upper_case) 
print(number) 
print(special_character)  

if length >= 8 and lower_case >= 1 and upper_case >= 1 and number >= 1 and special_character >= 1:
    print(password+" meet the criteria")
else:
    print(password+" failed. Choose another password")    


```
