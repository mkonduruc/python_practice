```python
# Question 38: Check if string is a valid palindrome ignoring case: "A man a plan a canal Panama"

input_value = "A man a plan a canal Panama"

input_transform = input_value.lower().replace(" ", "")

is_palindrome = True

for index in range(len(input_transform)):
    if(input_transform[index]!=input_transform[len(input_transform)-index-1]):
        is_palindrome = False

if is_palindrome:
     print(f"String '{input_value}' is a valid palindrome")
else:
     print(f"String '{input_value}' is not a valid palindrome")     
       
```

```python
# Question 39: Find the most common word in "the quick brown fox jumps over the lazy dog"

input_value = "the quick brown fox jumps over the lazy dog"
input_split = input_value.split()

word_count = 0
common_word = ''

for word in input_split:
    if input_value.count(word) > word_count:
        word_count = input_value.count(word)
        common_word = word

print(f"the most common word in '{input_value}' is '{common_word}'")
```

```python
# Question 40: Generate acronym from "National Aeronautics and Space Administration"

input_value = 'National Aeronautics and Space Administration'
input_split = input_value.split()
acronym =''

for word in input_split:
    if word not in ('and', 'of', 'in', 'the'):
        acronym += word[0]

print(f"acronym for '{input_value}' is '{acronym}'")
```

```python
# Question 41: Check if string contains balanced parentheses: "((()))"

parantheses_dict = {'(': ')', '[': ']', '{': '}'}

input_value = "({[]})"
parantheses_order = ""
is_balanced = True
input_char = ''

for word in input_value:
    if word in ("(",")","[","]","{","}"):
        parantheses_order += word

parantheses_length = len(parantheses_order)

for symbol_position in range(parantheses_length):
    input_char = parantheses_order[symbol_position] 
    if parantheses_dict[input_char] == parantheses_order[parantheses_length-symbol_position-1]:
        is_balanced = True
    else:
        is_balanced = False
        break

if is_balanced:
    print(f"given value {input_value} containes balanced parentheses") 
else:
    print(f"given value {input_value} doesnot containes balanced parentheses")            

```

```python
# Question 43: Find the longest common substring between "programming" and "grammar"

input_value1 = 'programming'
input_value2 = 'grammar'

common_substring = ''

if len(input_value1) > len(input_value2):
    compare_value = input_value2
    compare_against = input_value1
else:
    compare_value = input_value1
    compare_against = input_value2

for start_count in range(len(compare_value)):
    for repeat_count in range(len(compare_value)+1):
        if repeat_count>start_count and compare_value[start_count:repeat_count] in compare_against and len(common_substring) < len(compare_value[start_count:repeat_count]):
            common_substring = compare_value[start_count:repeat_count]  

print(f"the longest common substring between '{input_value1}' and '{input_value2}' is '{common_substring}")
```

```python
# Question 44: Check if string is a valid URL: "https://www.google.com"

input_value = "https://www.google.com"

input_split = input_value.split(".")

valid_protocol = False
valid_domain= 0
valid_formate= True

for splits in input_split:
    if splits == input_split[0] and (splits.startswith("http://") or splits.startswith("https://") or splits.startswith("ftp://")):
        valid_protocol = True
    else:
        if splits.isalnum():
            valid_domain += 1
        else: 
            if splits != input_split[1]:
               valid_domain += 1 
        if len(splits) == 0 or not(splits[0].isalnum()):
            valid_formate = False


if valid_domain == len(input_split)-1 and valid_formate == True and valid_protocol == True:
    print(f"Given URL {input_value} is valid")
else:
    print(f"Given URL {input_value} is not valid") 
```

```python
# Question 46: Convert "hello world" to Pig Latin

input_value = 'hello world'

input_split = input_value.split()
pig_latin = ""

for words in input_split:
    if words[0] in ('a','e','i','o','u'):
        pig_latin += words+'ay '
    else:
        pig_latin += words[1:]+words[0]+'ay '

print(f"Pig latin for {input_value} is {pig_latin}")
```

```python
# Question 47: Check if string is a valid IPv4 address: "192.168.1.1"

input_value = '192.168.1.1'

input_split = input_value.split('.')
is_valid = 0

if len(input_split) == 4:
    for inputs in input_split:
        if inputs.isnumeric() and 0 <= int(inputs) <= 255:
            is_valid += 1

if is_valid == 4:
    print(f"Given IPv4 address {input_value} is valid")
else:
    print(f"Given IPv4 address {input_value} is not valid")
```

```python
# Question 49: Convert "hello world" to ROT13 encoding

print(ord("A")) #65
print(ord("Z")) #90
print(ord("a")) #97
print(ord("z")) #122


input_value = "hello world"
rot13_value = ''

for letter in input_value:
    if letter.isalpha():
        if ('A' <= letter <= 'Z' and 65 <= ord(letter)+13 <=90) or ('a' <= letter <= 'z' and 97 <= ord(letter)+13 <=122):
            rot13_value += chr(ord(letter)+13)
        else:
            rot13_value += chr(ord(letter)-13) #ord(letter)+13-26 is simplified
    else:
         rot13_value += letter

print(f"ROT13 encoding of '{input_value}' is '{rot13_value}'")
    
```
