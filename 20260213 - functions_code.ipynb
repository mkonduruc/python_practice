# 1. Even or Odd

## Logic
- Take a number as input.
- Divide the number by 2.
- Check the remainder using modulus operator `%`.
- If remainder is 0 → number is Even.
- Otherwise → number is Odd.

```python
def even_odd(input_value):
    if input_value%2 == 0:
        value = 'Even'
    else:
        value = 'Odd'
    return value

input_value = int(input("Enter a number value to find it is even or odd: "))

print(f"Entered number '{input_value}' is {even_odd(input_value)} number")
```

# 2. Factorial of a Number

## Logic
- Take a number `n`.
- Start with a variable `result = 1`.
- Multiply `result` by all numbers from 1 to `n`.
- Return the final value of `result`.
- Special case: factorial of 0 is 1.

**Example:**
5! = 1 × 2 × 3 × 4 × 5

```python
def factorial_logic(input_value):
    factorial_value = 1
    for num in range(input_value,0,-1):
        factorial_value *= num
    return factorial_value

input_value = 5

print(f"Factorial of {input_value}: {factorial_logic(input_value)}")
```

# 3. Reverse a String

## Logic

### Method 1 (Loop)
- Take a string.
- Start from the last character.
- Add each character to a new string.
- Return the new reversed string.

### Method 2
- Use slicing with negative step.

```python
def reverse_string(input_string):
    reverse_string = ''
    for char in range(len(input_string),0,-1):
        reverse_string += input_string[char-1]
    return reverse_string

input_value = 'Python'

print(f"reverse string of '{input_value}': {reverse_string(input_value)}")
```

# 5. Largest Number in List

## Logic
- Assume first element is largest.
- Compare it with each element in the list.
- If a bigger number is found → update largest.
- Return the largest value.

```python
def max_number(input_list):
    max_value = 0
    for number in input_list:
        if number > max_value:
            max_value = number
    return max_value

input_list = [5,9,2,6,7,1]

print(f"largest number in {input_list}: {max_number(input_list)}")
```

# 6. Sum of List (without `sum()`)

## Logic
- Start with `total = 0`.
- Loop through each element.
- Add each element to `total`.
- Return `total`.

```python
def sum_of_list(input_list):
    total = 0
    for number in input_list:
        total += number
    return total

input_list = [1,2,3,4,5,6]

print(f"Sum of the list {input_list}: {sum_of_list(input_list)}")
```

# 7. Count Vowels

## Logic
- Create a string of vowels (`aeiouAEIOU`).
- Set `count = 0`.
- Loop through each character.
- If character is in vowels → increase `count`.
- Return `count`.

```python
def count_vowels(input_value):
    count=0
    for char in input_value:
        if char.lower() in ('a','e','i','o','u'):
            count += 1
    return count

input_value = 'aeiouAEIO'

print(f"Number of vowels in '{input_value}' : {count_vowels(input_value)}")
```

# 8. Fibonacci Sequence

## Logic
- First two numbers are 0 and 1.
- Next number = sum of previous two.
- Repeat until required `n` terms.
- Store numbers in a list.
- Return the list.

**Example:**
0, 1, 1, 2, 3, 5, 8…

```python

def fibnocci_series(input_value):
    previous_number = 0
    latest_number = 0
    print(f"Fibonacci numbers series for {input_value} :",end=' ')
    for number in range(0,input_value+1,):
        if number == 0:
            previous_number = 0
            latest_number = 1
            print(f"{previous_number} , {latest_number}", end=', ')
        else:
            value_holder = previous_number
            previous_number = latest_number
            latest_number += value_holder 
            print(latest_number, end=', ')


input_value = 10

fibnocci_series(input_value)       
```

# 9. Palindrome Check

## Logic
- Take a string.
- Reverse the string.
- Compare original string with reversed string.
- If same → Palindrome.
- Else → Not palindrome.

**Example:**
"madam" → same when reversed

```python
def palindrome_check(input_value):
    reversed_value = ''
    for numb in range(len(input_value),0,-1):
        reversed_value += input_value[numb-1]
    if input_value == reversed_value:
        is_palindrom = 'is Palindrome'
    else:
        is_palindrom = 'is not Palindrome'
    return is_palindrom    

input_value = 'madam'

print(f"{input_value} {palindrome_check(input_value)}")
```

# 10. GCD (Greatest Common Divisor)

## Logic (Euclidean Method)
- Take two numbers `a` and `b`.
- While `b` is not 0:
  - Replace `a` with `b`.
  - Replace `b` with remainder of `a % b`.
- When `b` becomes 0 → `a` is GCD.

```python
def gcd_caliculation(a,b):
    while b != 0:
        gcd = b
        b = a%b
        a = gcd
    return a

print(f"GCD of 12, 15 is : {gcd_caliculation(12, 15)}")
```

# 11. Remove Duplicates from List

## Logic
- Create empty list.
- Loop through original list.
- If element is not already in new list → add it.
- Return new list.

```python
def remove_duplicates(input_list):
    output_list = []
    for input in input_list:
        if input not in output_list:
            output_list.append(input)
    return output_list

input_list = [5,2,4,3,1,2,9,3,1]

print(f"List after removing duplicates from {input_list}: {remove_duplicates(input_list)}")
```

# 12. Get Even Numbers from List

## Logic
- Create empty list.
- Loop through original list.
- If number `% 2 == 0` → add to new list.
- Return new list.

```python
def even_numbers(input_list):
    even_list = []
    for number in input_list:
        if number % 2 == 0:
            even_list.append(number)
    return even_list

input_list = [5,2,4,6,7,8,3]

print(f"Even numbers from {input_list}: {even_numbers(input_list)}")
```

# 13. Character Frequency in String

## Logic
- Create empty dictionary.
- Loop through each character.
- If character already in dictionary:
  - Increase its count.
- Else:
  - Add character with count 1.
- Return dictionary.

```python
def character_frequency(string):
    frequency_dict = {}
    for char in string:
        if frequency_dict.get(char,0) == 0:
            frequency_dict[char] = 1
        else:
            frequency_dict[char] += 1
    return frequency_dict        

input_value = 'example'

print(f"character frequency in '{input_value}': {character_frequency(input_value)}")

```

# 14. Swap Two Numbers (Without Third Variable)

## Logic

### Method 1 (Python way)
`a, b = b, a`

### Method 2 (Math way)
- `a = a + b`
- `b = a - b`
- `a = a - b`


```python
def swap_numbers(a,b):
    a, b = b, a
    return a, b

print(f"swap number '5 , 7' : {swap_numbers(5, 7)}")
```

# 15. Celsius to Fahrenheit

## Logic
- Take temperature in Celsius.
- Use formula:
  `Fahrenheit = (Celsius × 9/5) + 32`
- Return the result.

```python
def convert_temperature(Celsius):
    Fahrenheit = (Celsius * (9/5)) + 32
    return Fahrenheit 

input_value = float(input("Enter temperature in Celsius to convet Fahrenheit"))

print(f"{input_value} Celsius = {convert_temperature(input_value)} Fahrenheit")
```
